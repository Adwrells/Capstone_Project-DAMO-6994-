```python
import pandas as pd
from pathlib import Path
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Border, Side, Alignment
from openpyxl.utils import get_column_letter
```

```python
# ============================================================
# INPUT & OUTPUT PATHS
# ============================================================

INPUT_FOLDER = Path(
    r"C:\Users\bhara\OneDrive\Desktop\DAMO-699-Capstone-Project-Unofficial-\DAMO-699-Capstone-Project-Unofficial-\data\Explorer Dataset"
)

OUTPUT_FOLDER = Path(
    r"C:\Users\bhara\OneDrive\Desktop\DAMO-699-Capstone-Project-Unofficial-\DAMO-699-Capstone-Project-Unofficial-\data\cleaned dataset"
)

OUTPUT_FOLDER.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = OUTPUT_FOLDER / "Explanatory_and_Predictive_ED_Analytics_Dataset.xlsx"
```

```python
datasets = {

    "Age_Sex": pd.read_excel(INPUT_FOLDER / "Age_Sex_Preview.xlsx"),

    "CTAS_Triage": pd.read_csv(INPUT_FOLDER / "CTAS_Triage.csv"),

    "Visit_Disposition": pd.read_csv(INPUT_FOLDER / "Visit_Disposition.csv"),

    "ED_Visits": pd.read_csv(INPUT_FOLDER / "ED_Visits.csv"),

    "Main_Problems": pd.read_csv(INPUT_FOLDER / "Main_Problems.csv"),

    "Demographics": pd.read_csv(INPUT_FOLDER / "Demographics.csv")

}
```

```python
readme = pd.DataFrame({

    "Information":[

        "Explanatory & Predictive Emergency Department Analytics Dataset",

        "",

        "University of Niagara Falls",

        "Master of Data Analytics",

        "",

        "Healthcare Capstone Project",

        "",

        "Prepared by Bharath Paramasivan",

        "",

        "This workbook contains cleaned and standardized datasets.",

        "Designed for exploratory analysis, hypothesis testing,",

        "predictive analytics and executive dashboards.",

        "",

        "Workbook Contents",

        "",

        "README",

        "Master_Metadata",

        "Data_Quality_Report",

        "Data_Dictionary",

        "Age_Sex",

        "CTAS_Triage",

        "Visit_Disposition",

        "ED_Visits",

        "Main_Problems",

        "Demographics"

    ]

})
```

```python
metadata = []

for name, df in datasets.items():

    metadata.append({

        "Dataset": name,

        "Rows": len(df),

        "Columns": len(df.columns),

        "Missing Values": int(df.isnull().sum().sum()),

        "Duplicate Rows": int(df.duplicated().sum())

    })

metadata = pd.DataFrame(metadata)
```

```python
quality = []

for name, df in datasets.items():

    quality.append({

        "Dataset": name,

        "Rows": len(df),

        "Columns": len(df.columns),

        "Missing Values": int(df.isnull().sum().sum()),

        "Duplicate Rows": int(df.duplicated().sum()),

        "Memory (KB)": round(df.memory_usage(deep=True).sum()/1024,2)

    })

quality = pd.DataFrame(quality)
```

```python
dictionary = []

for dataset_name, df in datasets.items():

    for column in df.columns:

        dictionary.append({

            "Dataset": dataset_name,

            "Column Name": column,

            "Data Type": str(df[column].dtype),

            "Missing Values": int(df[column].isnull().sum()),

            "Unique Values": int(df[column].nunique())

        })

dictionary = pd.DataFrame(dictionary)
```

```python
with pd.ExcelWriter(
    OUTPUT_FILE,
    engine="openpyxl"
) as writer:

    readme.to_excel(writer,
                    sheet_name="README",
                    index=False)

    metadata.to_excel(writer,
                      sheet_name="Master_Metadata",
                      index=False)

    quality.to_excel(writer,
                     sheet_name="Data_Quality_Report",
                     index=False)

    dictionary.to_excel(writer,
                        sheet_name="Data_Dictionary",
                        index=False)

    for sheet_name, df in datasets.items():

        df.to_excel(
            writer,
            sheet_name=sheet_name,
            index=False
        )
```

```python
wb = load_workbook(OUTPUT_FILE)

header_fill = PatternFill(
    fill_type="solid",
    start_color="1F4E78",
    end_color="1F4E78"
)

header_font = Font(
    color="FFFFFF",
    bold=True
)

thin = Side(style="thin")

border = Border(
    left=thin,
    right=thin,
    top=thin,
    bottom=thin
)

for ws in wb.worksheets:

    ws.freeze_panes = "A2"

    ws.auto_filter.ref = ws.dimensions

    for cell in ws[1]:

        cell.fill = header_fill
        cell.font = header_font
        cell.alignment = Alignment(
            horizontal="center",
            vertical="center"
        )
        cell.border = border

    for column_cells in ws.columns:

        max_length = max(
            len(str(cell.value)) if cell.value is not None else 0
            for cell in column_cells
        )

        adjusted_width = min(max_length + 3, 40)

        ws.column_dimensions[
            get_column_letter(column_cells[0].column)
        ].width = adjusted_width

wb.save(OUTPUT_FILE)

print("="*80)
print("SUCCESS!")
print("="*80)

print(f"Workbook saved to:\n{OUTPUT_FILE}")

print("\nSheets Created:")

for sheet in wb.sheetnames:

    print(f"✓ {sheet}")

print("\nWorkbook completed successfully.")
```
