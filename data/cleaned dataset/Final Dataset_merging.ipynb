```python
from pathlib import Path
import pandas as pd
```

```python
INPUT_FOLDER = Path(
    r"C:\Users\bhara\OneDrive\Desktop\DAMO-699-Capstone-Project-Unofficial-\DAMO-699-Capstone-Project-Unofficial-\data\optimized"
)
```

```python
OUTPUT_FOLDER = Path(
    r"C:\Users\bhara\OneDrive\Desktop\DAMO-699-Capstone-Project-Unofficial-\DAMO-699-Capstone-Project-Unofficial-\data\cleaned dataset"
)
```

```python
OUTPUT_FOLDER.mkdir(parents=True, exist_ok=True)
```

```python
OUTPUT_FILE = OUTPUT_FOLDER / "Explanatory_and_Predictive_ED_Analytics_Dataset.xlsx"
```

```python
ED_VISITS_FILE = INPUT_FOLDER / "ED Visits from 2003 - 2021.csv"
DEMOGRAPHICS_FILE = INPUT_FOLDER / "ED visits by month, age and sex, participating provinces.csv"
TOP10_FILE = INPUT_FOLDER / "Top 10 Main Problems.csv"
```

```python
print("=" * 60)
print("Loading datasets...")
print("=" * 60)
```

```python
ed_visits = pd.read_csv(ED_VISITS_FILE)
```

```python
demographics = pd.read_csv(DEMOGRAPHICS_FILE)
```

```python
top10 = pd.read_csv(TOP10_FILE)
```

```python
print("ED Visits loaded")
print("Demographics loaded")
print("Top 10 Main Problems loaded")
```

```python
print("\nCreating Excel workbook...")

with pd.ExcelWriter(
    OUTPUT_FILE,
    engine="openpyxl"
) as writer:

    ed_visits.to_excel(
        writer,
        sheet_name="ED_Visits_2003_2021",
        index=False
    )

    demographics.to_excel(
        writer,
        sheet_name="Demographics",
        index=False
    )

    top10.to_excel(
        writer,
        sheet_name="Top10_Main_Problems",
        index=False
    )
```

```python
print("\n" + "=" * 60)
print("Workbook created successfully!")
print("=" * 60)
```

```python
print(f"Saved to:\n{OUTPUT_FILE}")
```
