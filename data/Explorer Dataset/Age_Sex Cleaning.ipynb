```python
print("\nRemoving rows where Sex = 'All'...")

df = df[df["sex"].str.lower() != "all"].copy()
```

```python
df.rename(
    columns={
        "median_length_of_stay_min": "median_los_minutes",
        "length_of_stay_hours": "median_los_hours"
    },
    inplace=True
)
```

```python
df["median_los_minutes"] = (
    pd.to_numeric(df["median_los_minutes"], errors="coerce")
    .round()
    .astype("Int64")
)
```

```python
def population_category(age):

    age = str(age).strip()

    pediatric = [
        "0-4", "5-9", "10-14", "15-19",
        "00-04", "05-09", "10-14", "15-19"
    ]

    young_adult = [
        "20-24", "25-29", "30-34"
    ]

    adult = [
        "35-39", "40-44", "45-49"
    ]

    pre_senior = [
        "50-54", "55-59", "60-64"
    ]

    geriatric = [
        "65-69", "70-74", "75-79", "80-84", "85+"
    ]

    if age in pediatric:
        return "0–17: Pediatric Population"

    elif age in young_adult:
        return "18–34: Young Adult Population"

    elif age in adult:
        return "35–49: Adult Population"

    elif age in pre_senior:
        return "50–64: Pre-Senior Population"

    elif age in geriatric:
        return "65–85+: Geriatric Population"

    else:
        return "Unknown"


df["population_category"] = df["age_group"].apply(population_category)
```

```python
if "age_broad_category" in df.columns:
    df.drop(columns=["age_broad_category"], inplace=True)
```

```python
df = df[
    [
        "fiscal_year",
        "fiscal_year_start",
        "sex",
        "age_group",
        "population_category",
        "ed_visits",
        "median_los_minutes",
        "median_los_hours",
    ]
]
```

```python
df = df.sort_values(
    by=[
        "fiscal_year_start",
        "age_group",
        "sex"
    ]
).reset_index(drop=True)
```

```python
# Normalize age group formatting
df["age_group"] = (
    df["age_group"]
    .astype(str)
    .str.strip()
    .str.replace("–", "-", regex=False)
    .str.replace("—", "-", regex=False)
)
```

```python
age_mapping = {
    "00-04": "0–17",
    "0-4": "0–17",
    "05-09": "0–17",
    "5-9": "0–17",
    "10-14": "0–17",
    "15-19": "0–17",

    "20-24": "18–34",
    "25-29": "18–34",
    "30-34": "18–34",

    "35-39": "35–49",
    "40-44": "35–49",
    "45-49": "35–49",

    "50-54": "50–64",
    "55-59": "50–64",
    "60-64": "50–64",

    "65-69": "65–85+",
    "70-74": "65–85+",
    "75-79": "65–85+",
    "80-84": "65–85+",
    "85+": "65–85+"
}
```

```python
df["age_group"] = df["age_group"].map(age_mapping)
```

```python
population_mapping = {
    "0–17": "Pediatric Population",
    "18–34": "Young Adult Population",
    "35–49": "Adult Population",
    "50–64": "Pre-Senior Population",
    "65–85+": "Geriatric Population"
}

df["population_category"] = df["age_group"].map(population_mapping)
```

```python
# Weighted average function
def weighted_average(group):
    return round(
        np.average(
            group["median_los_minutes"],
            weights=group["ed_visits"]
        )
    )

df = (
    df.groupby(
        [
            "fiscal_year",
            "fiscal_year_start",
            "sex",
            "age_group",
            "population_category"
        ],
        as_index=False
    )
    .apply(
        lambda g: pd.Series({
            "ed_visits": g["ed_visits"].sum(),
            "median_los_minutes": weighted_average(g)
        })
    )
    .reset_index(drop=True)
)
```

```python
df["median_los_hours"] = (
    df["median_los_minutes"] / 60
).round(2)
```

```python
df = df[
    [
        "fiscal_year",
        "fiscal_year_start",
        "sex",
        "age_group",
        "population_category",
        "ed_visits",
        "median_los_minutes",
        "median_los_hours"
    ]
]
```

```python
age_order = {
    "0-17": 1,
    "18-34": 2,
    "35-49": 3,
    "50-64": 4,
    "65-85+": 5
}

df["age_sort"] = df["age_group"].map(age_order)

df = (
    df.sort_values(
        ["fiscal_year_start", "age_sort", "sex"]
    )
    .drop(columns="age_sort")
    .reset_index(drop=True)
)
```
