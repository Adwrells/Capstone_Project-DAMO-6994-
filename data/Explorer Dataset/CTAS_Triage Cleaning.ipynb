```python
print("\nRemoving 'All' from Sex...")

df["sex"] = (
    df["sex"]
    .astype(str)
    .str.strip()
)

df = df[df["sex"].str.lower() != "all"].copy()

print("Remaining Sex Categories")
print(df["sex"].value_counts())
```

```python
print("\nRemoving 'Total' Age Group...")

df["age_group"] = (
    df["age_group"]
    .astype(str)
    .str.strip()
)

df = df[df["age_group"].str.lower() != "total"].copy()

print(df["age_group"].value_counts())
```

```python
print("\nRenaming Columns...")

df.rename(
    columns={
        "median_length_of_stay_min": "median_los_minutes",
        "length_of_stay_hours": "median_los_hours",
        "age_broad_category": "population_category"
    },
    inplace=True
)

print(df.columns.tolist())
```

```python
print("\nConverting Median LOS to Whole Numbers...")

df["median_los_minutes"] = (
    pd.to_numeric(
        df["median_los_minutes"],
        errors="coerce"
    )
    .round(0)
    .astype(int)
)
```

```python
df["median_los_hours"] = (
    df["median_los_minutes"] / 60
).round(2)
```

```python
print("\nStandardizing CTAS Levels...")

ctas_mapping = {

    "Resuscitation":"CTAS I - Resuscitation",

    "Emergent":"CTAS II - Emergent",

    "Urgent":"CTAS III - Urgent",

    "Less Urgent":"CTAS IV - Less Urgent",

    "Non Urgent":"CTAS V - Non-Urgent",

    "Non-Urgent":"CTAS V - Non-Urgent"

}

df["triage_level"] = (
    df["triage_level"]
    .astype(str)
    .str.strip()
    .replace(ctas_mapping)
)

print(df["triage_level"].unique())
```

```python
print("\nCreating Population Category...")

df["population_category"] = np.where(
    df["age_group"].str.contains("00", na=False),
    "Pediatric & Youth",

    np.where(
        df["age_group"].str.contains("20", na=False),
        "Young Adult",

        np.where(
            df["age_group"].str.contains("45", na=False),
            "Middle Adult",

            np.where(
                df["age_group"].str.contains("65", na=False),
                "Older Adult",
                "Unknown"
            )
        )
    )
)
```

```python
print("\nAge Groups")
print(df["age_group"].value_counts())

print("\nPopulation Categories")
print(df["population_category"].value_counts())

print("\nCTAS Levels")
print(df["triage_level"].value_counts())

print("\nMedian LOS")
print(df["median_los_minutes"].describe())

print("\nPreview")

display(df.head(20))
```

```python
print("\nSorting Dataset...")

df = (
    df.sort_values(
        by=[
            "fiscal_year_start",
            "ctas_urgency_score",
            "sex",
            "age_group"
        ],
        ascending=[False, True, True, True]
    )
    .reset_index(drop=True)
)

print("Dataset Sorted Successfully.")
```

```python
print("\nReordering Columns...")

final_columns = [

    "fiscal_year",

    "fiscal_year_start",

    "sex",

    "triage_level",

    "ctas_urgency_score",

    "age_group",

    "population_category",

    "ed_visits",

    "median_los_minutes",

    "median_los_hours"

]

df = df[final_columns]
```
