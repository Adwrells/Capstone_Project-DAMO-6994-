```python
df["sex"] = (
    df["sex"]
    .astype(str)
    .str.strip()
    .str.title()
)

df = df[df["sex"] != "All"].copy()
```

```python
df["age_group"] = (
    df["age_group"]
    .astype(str)
    .str.strip()
    .str.replace("–", "-", regex=False)
)

df = df[df["age_group"] != "Total"].copy()
```

```python
age_mapping = {

    "00-04":"0-19",
    "05-09":"0-19",
    "10-14":"0-19",
    "15-19":"0-19",

    "20-24":"20-44",
    "25-29":"20-44",
    "30-34":"20-44",
    "35-39":"20-44",
    "40-44":"20-44",

    "45-49":"45-64",
    "50-54":"45-64",
    "55-59":"45-64",
    "60-64":"45-64",

    "65-69":"65+",
    "70-74":"65+",
    "75-79":"65+",
    "80-84":"65+",
    "85+":"65+"

}
```

```python
df["age_group"] = df["age_group"].replace(age_mapping)
```

```python
population_mapping = {

    "0-19":"Pediatric & Youth",

    "20-44":"Young Adult",

    "45-64":"Middle Adult",

    "65+":"Older Adult"

}

df["population_category"] = df["age_group"].map(population_mapping)
```

```python
df.rename(
    columns={
        "median_length_of_stay_min":"median_los_minutes",
        "length_of_stay_hours":"median_los_hours",
        "is_admitted":"admission_flag"
    },
    inplace=True
)
```

```python
df["visit_disposition"] = (
    df["visit_disposition"]
    .astype(str)
    .str.strip()
    .replace({

        "Discharged home":"Discharged",

        "Not seen or left":"Left Without Being Seen",

        "Transferred":"Transferred",

        "Intra-facility transfer":"Intra-Facility Transfer",

        "Admitted":"Admitted",

        "Death":"Death",

        "Unknown":"Unknown"

    })
)
```

```python
df["admission_status"] = df["admission_flag"].map({

    0:"Not Admitted",

    1:"Admitted"

})
```

```python
df["median_los_minutes"] = (
    pd.to_numeric(
        df["median_los_minutes"],
        errors="coerce"
    )
    .round()
    .astype(int)
)

df["median_los_hours"] = (
    df["median_los_minutes"]/60
).round(2)
```

```python
df["ed_visits"] = (
    pd.to_numeric(
        df["ed_visits"],
        errors="coerce"
    )
    .fillna(0)
    .astype(int)
)
```

```python
df["erbi"] = (
    df["ed_visits"] *
    df["median_los_minutes"]
)
```

```python
print("\nSex")
print(df["sex"].value_counts())

print("\nAge Groups")
print(df["age_group"].value_counts())

print("\nPopulation Categories")
print(df["population_category"].value_counts())

print("\nVisit Disposition")
print(df["visit_disposition"].value_counts())

print("\nAdmission Status")
print(df["admission_status"].value_counts())

display(df.head(20))
```
