```python
from IPython.display import display
import pandas as pd
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
```

```python
INPUT_FILE = Path(
    r"C:\Users\bhara\OneDrive\Desktop\DAMO-699-Capstone-Project-Unofficial-\DAMO-699-Capstone-Project-Unofficial-\data\optimized\ED_Visits.csv"
)

OUTPUT_FOLDER = Path(
    r"C:\Users\bhara\OneDrive\Desktop\DAMO-699-Capstone-Project-Unofficial-\DAMO-699-Capstone-Project-Unofficial-\data\Explorer Dataset"
)

OUTPUT_FOLDER.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = OUTPUT_FOLDER / "ED_Visits.csv"
```

```python
df = pd.read_csv(INPUT_FILE)

print("=" * 90)
print("ED VISITS DATASET LOADED")
print("=" * 90)

display(df.head())
```

```python
print("\nDataset Shape")
print(df.shape)

print("\nColumn Names")
print(df.columns.tolist())

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Rows")
print(df.duplicated().sum())
```

```python
df.rename(
    columns={
        "median_length_of_stay_min": "median_los_minutes",
        "length_of_stay_hours": "median_los_hours",
        "is_admitted": "admission_status"
    },
    inplace=True
)
```

```python
ctas_mapping = {

    "Resuscitation": "CTAS I - Resuscitation",

    "Emergent": "CTAS II - Emergent",

    "Urgent": "CTAS III - Urgent",

    "Less Urgent": "CTAS IV - Less Urgent",

    "Non Urgent": "CTAS V - Non-Urgent",

    "Non-Urgent": "CTAS V - Non-Urgent"

}

df["triage_level"] = (
    df["triage_level"]
      .astype(str)
      .str.strip()
      .replace(ctas_mapping)
)
```

```python
df["admission_status"] = (
    df["admission_status"]
      .replace({
          0: "Not Admitted",
          1: "Admitted"
      })
)
```

```python
df["median_los_minutes"] = (
    pd.to_numeric(
        df["median_los_minutes"],
        errors="coerce"
    )
    .round()
    .astype(int)
)

df["median_los_hours"] = (
    df["median_los_minutes"] / 60
).round(2)
```

```python
df = (
    df.sort_values(
        by=[
            "fiscal_year_start",
            "ctas_urgency_score",
            "admission_status",
            "visit_disposition",
            "main_problem"
        ],
        ascending=[
            False,
            True,
            True,
            True,
            True
        ]
    )
    .reset_index(drop=True)
)
```

```python
df = df[
    [
        "fiscal_year",
        "fiscal_year_start",
        "triage_level",
        "ctas_urgency_score",
        "visit_disposition",
        "main_problem",
        "admission_status",
        "ed_visits",
        "median_los_minutes",
        "median_los_hours"
    ]
]
```

```python
print("\nRows")
print(len(df))

print("\nColumns")
print(len(df.columns))

print("\nColumn Names")
print(df.columns.tolist())

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Rows")
print(df.duplicated().sum())

print("\nCTAS Levels")
print(df["triage_level"].unique())

print("\nAdmission Status")
print(df["admission_status"].value_counts())

print("\nFiscal Years")
print(df["fiscal_year"].unique())

print("\nVisit Disposition")
print(df["visit_disposition"].unique())

print("\nMain Problems")
print(df["main_problem"].nunique())
```

```python
print("\nFirst 20 Rows")
display(df.head(20))

print("\nLast 20 Rows")
display(df.tail(20))
```

```python
df["triage_level"] = (
    df["triage_level"]
    .astype(str)
    .str.strip()
)

# Remove rows where triage_level = Total
df = df[
    df["triage_level"].str.lower() != "total"
].copy()
```

```python
print(df["triage_level"].unique())

print("\nValue Counts")
print(df["triage_level"].value_counts())

print(f"\nTotal Rows Remaining: {len(df):,}")
```

```python
# ============================================================
# STANDARDIZE VISIT DISPOSITION
# ============================================================

df["visit_disposition"] = (
    df["visit_disposition"]
    .astype(str)
    .str.strip()
    .str.title()
)

visit_disposition_mapping = {

    "Discharged":"Discharged",

    "Admitted":"Admitted",

    "Transferred":"Transferred",

    "Left Without Being Seen":"Left Without Being Seen",

    "Left Against Medical Advice":"Left Against Medical Advice",

    "Died":"Died"

}

df["visit_disposition"] = (
    df["visit_disposition"]
    .replace(visit_disposition_mapping)
)

print(df["visit_disposition"].value_counts())
```

```python
df["main_problem"] = (
    df["main_problem"]
    .astype(str)
    .str.strip()
    .str.title()
    .str.replace(r"\s+", " ", regex=True)
)

print(df["main_problem"].nunique())
```

```python
df["ed_visits"] = (
    pd.to_numeric(
        df["ed_visits"],
        errors="coerce"
    )
    .fillna(0)
    .astype("int64")
)

print(df["ed_visits"].dtype)
```

```python
df["admission_flag"] = (
    df["admission_status"]
    .map({
        "Not Admitted":0,
        "Admitted":1
    })
    .astype(int)
)

display(
    df[
        [
            "admission_status",
            "admission_flag"
        ]
    ].head()
)
```

```python
duplicate_keys = [

    "fiscal_year",

    "triage_level",

    "visit_disposition",

    "main_problem",

    "admission_status"

]

duplicates = df[
    df.duplicated(
        subset=duplicate_keys,
        keep=False
    )
]

print("Duplicate Analytical Records")

print(len(duplicates))

display(
    duplicates.sort_values(
        duplicate_keys
    ).head(50)
)
```

```python
if len(duplicates) > 0:

    df = (

        df.groupby(

            duplicate_keys,

            as_index=False

        )

        .agg({

            "ed_visits":"sum",

            "median_los_minutes":"median",

            "median_los_hours":"median",

            "ctas_urgency_score":"first",

            "fiscal_year_start":"first",

            "admission_flag":"first"

        })

    )

print(df.shape)
```

```python
df["erbi"] = df["ed_visits"] * df["median_los_minutes"]
```

```python
# Remove rows where visit_disposition is "Total"
df = df[df["visit_disposition"].str.strip().str.lower() != "total"]

# Remove rows where main_problem is "Any"
df = df[df["main_problem"].str.strip().str.lower() != "any"]

# Reset row index
df = df.reset_index(drop=True)
```

```python
print("\nMain Problem values:")
print(df["main_problem"].unique())
```
