```python
print("\nRemoving 'All' from Sex...")

df["sex"] = (
    df["sex"]
    .astype(str)
    .str.strip()
    .str.title()
)

df = df[
    df["sex"] != "All"
].copy()

print(df["sex"].value_counts())
```

```python
print("\nRemoving Total Age Group...")

df["age_group"] = (
    df["age_group"]
    .astype(str)
    .str.strip()
    .str.replace("–","-",regex=False)
)

df = df[
    df["age_group"] != "Total"
].copy()
```

```python
age_mapping = {

    "00-04":"0-19",
    "05-09":"0-19",
    "10-14":"0-19",
    "15-19":"0-19",

    "20-24":"20-44",
    "25-29":"20-44",
    "30-34":"20-44",
    "35-39":"20-44",
    "40-44":"20-44",

    "45-49":"45-64",
    "50-54":"45-64",
    "55-59":"45-64",
    "60-64":"45-64",

    "65-69":"65+",
    "70-74":"65+",
    "75-79":"65+",
    "80-84":"65+",
    "85+":"65+"

}

df["age_group"] = (
    df["age_group"]
    .replace(age_mapping)
)
```

```python
population_mapping = {

    "0-19":"Pediatric & Youth",

    "20-44":"Young Adult",

    "45-64":"Middle Adult",

    "65+":"Older Adult"

}

df["population_category"] = (
    df["age_group"]
    .replace(population_mapping)
)
```

```python
rename_columns = {

    "median_length_of_stay_min":"median_los_minutes",

    "length_of_stay_hours":"median_los_hours"

}

df.rename(
    columns=rename_columns,
    inplace=True
)
```

```python
df["median_los_minutes"] = (

    pd.to_numeric(
        df["median_los_minutes"],
        errors="coerce"
    )

    .round()

    .astype(int)

)

df["median_los_hours"] = (

    df["median_los_minutes"]/60

).round(2)
```

```python
print("\nCleaning Main Problem...")

df["main_problem"] = (

    df["main_problem"]

    .astype(str)

    .str.strip()

    .str.title()

    .str.replace(
        r"\s+",
        " ",
        regex=True
    )

)
```

```python
df["problem_rank"] = (

    df.groupby("fiscal_year")["ed_visits"]

      .rank(
          method="dense",
          ascending=False
      )

      .astype(int)

)
```

```python
df["erbi"] = (

    df["ed_visits"]

    *

    df["median_los_minutes"]

)
```

```python
print("VERIFICATION")
print("="*90)

print("\nSex")

print(df["sex"].value_counts())

print("\nAge Groups")

print(df["age_group"].value_counts())

print("\nPopulation Category")

print(df["population_category"].value_counts())

print("\nTop Problems")

display(
    df["main_problem"]
    .value_counts()
    .head(20)
)

print("\nProblem Rank")

display(
    df[
        [
            "main_problem",
            "problem_rank"
        ]
    ].head(20)
)

print("\nERBI Summary")

display(
    df["erbi"].describe()
)

display(df.head(20))
```
